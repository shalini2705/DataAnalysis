```python
import numpy as np
#1D
arr = np.array([10, 20, 30, 40, 50])
print(np.sum(arr))#sum
print(np.min(arr))#minimun value
print(np.max(arr))#max value
print(np.sum(arr,axis=0))
print(np.mean(arr))
print(np.var(arr))
print(np.std(arr))
print(np.percentile(arr,80))
print(np.quantile(arr,0.80))
print(np.quantile(arr,1))
#percentile->percen->0-100
#quantile->proportion->0-1
print(np.argmax(arr))#gives index value of max digit
print(np.argmin(arr))#gives index value of min digit
arr=np.array([10,100,1000])
print(np.log10(arr))#logs
print(np.log2(arr))#logs
print(np.log(arr))
print(np.log(2))
print(np.log(5))
arr=np.array([10.25,20.675,30.3])#rounding of numbers
print(np.round(arr))
arr=np.array([10.765,20.334,30.908])
print(np.round(arr,2))#2 reflects the decimal 2nd value
print(np.floor(arr))#gives lower limit values and decimal point before values
print(np.ceil(arr))#ives upper limit values and next number
arr=np.array([-10,-5,0,-3,6,4,30])
print(np.abs(arr))#removes -ve and gives +ve values
arr = np.array([780, 450, 870, 300, 70])
print(np.sort(arr))#sorting
print(np.sort(arr)[::-1])#decending order
print(np.sort(arr)[::1])#aesending order
print(arr[::-1])#reverse



```

```python
#2d
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])
print(np.sum(arr))
print(np.min(arr))
print(np.max(arr))
print(np.sum(arr,axis=0))
print(np.mean(arr))
print(np.var(arr))
print(np.std(arr))
print(np.percentile(arr,80))
print(np.quantile(arr,1))
print(np.argmax(arr))
print(np.argmin(arr))
print(np.log10(arr))
print(np.log2(arr))
print(np.log(arr))
```

```python
import numpy as np
arr1 = np.array([10, 20, 30])
arr2 = np.array([40, 50, 60])
result = np.concatenate((arr1, arr2))
print(result) #concatenation it combines 2 arrays into 1 array and changes it into 1D
print(result.ndim)
arr1 = np.array([
[10, 20],
[30, 40] ])
arr2 = np.array([
[50, 60],
[70, 80]])
result = np.concatenate((arr1, arr2), axis=1) #2d horizontal concatenation
print(result)
print(result.ndim)
result = np.concatenate((arr1, arr2), axis=0) #2d vertical concatenation
print(result)
print(result.ndim)
arr1 = np.array([
[10, 20],
[30, 40] ])
arr2 = np.array([
[50, 60],
[70, 80]])
result = np.stack((arr1, arr2), axis=0)#stack changes dimension
print(result)
print(result.ndim)
result = np.stack((arr1, arr2), axis=1)
print(result)
print(result.ndim)
arr1 = np.array([10, 20, 30])
arr2 = np.array([40, 50, 60])
result = np.stack((arr1, arr2),axis=0)
print(result)
result = np.stack((arr1, arr2),axis=1)
print(result)
arr1 = np.array([10, 20, 30]) #1d
arr2 = np.array([40, 50, 60])
result = np.vstack((arr1, arr2))#vstack exception case (chnages 1d to 2d)
print(result)
print(result.ndim)
arr1 = np.array([ #2d
[10, 20],
[30, 40] ])
arr2 = np.array([
[50, 60],
[70, 80]])
result = np.vstack((arr1, arr2))
print(result)
print(result.ndim)
#3d
arr1 = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])
arr2 = np.array([
    [[11,22], [33, 44]],
    [[55,66], [77, 88]]
])
result=np.vstack((arr1,arr2))
print(result)
print(result.ndim)
#h stack
arr1 = np.array([10, 20, 30]) #1d
arr2 = np.array([40, 50, 60])
result = np.hstack((arr1, arr2))
print(result)
print(result.ndim)

```
