```python
import pandas as pd
data = {
    "Name":["Ravi","Priya","Arun","Sneha","Kiran","Anjali","Vikram","Meena"],
    "Age":[22,24,21,25,23,26,20,27],
    "City":["Hyderabad","Chennai","Bangalore","Hyderabad","Mumbai","Chennai","Bengalore","Mumbai"],
    "Salary":[30000,40000,28000,50000,35000,45000,25000,55000]
}
df=pd.DataFrame(data)
print(df)
df.groupby('City')['Age'].agg(['mean','max','min'])
```

```python
df.groupby('City')['Age'].agg(['mean','max','min'])
```

```python
df.groupby('City').agg({'Age':['mean','max'],'Salary':['sum','count']})
```

```python
df1=pd.DataFrame({
    'ID':[1,2,3],
    'Name':['a','b','c']
})
df2=pd.DataFrame({
    'ID':[1,2,4],
    'Salary':[100,200,300]

})
print(df1)
print(df2)
```

```python
merged=pd.merge(
    df1,df2,on='ID',how='inner'
)
print(merged)

```

```python
merged=pd.merge(
    df1,df2,on='ID',how='left'
)
print(merged)
```

```python
merged=pd.merge(
    df1,df2,on='ID',how='right'
)
print(merged)
```

```python
merged=pd.merge(
    df1,df2,on='ID',how='outer'
)
print(merged)
```

```python
combined=pd.concat(
    [df1,df2],axis=0,ignore_index=True
)
print(combined)
```

```python
combined=pd.concat(
    [df1,df2],axis=1
)
print(combined)
```
