```python
import matplotlib.pyplot as plt#line chart
import numpy as np
products=["laptop","mobile","tablet","headphones","watch"]
sales=[120,180,90,150,110]
profit=[30,50,20,40,25]
plt.plot(products,sales,marker="*")
plt.title("products vs sales")
plt.xlabel("products")
plt.ylabel("sales")
plt.show()
```

```python
plt.plot(products,sales,marker="*",label="sales")#multiple line chart
plt.plot(products,profit,marker="o",label="profit")
plt.title("products vs sales")
plt.xlabel("products")
plt.ylabel("amount")
plt.legend()#used to show legend
plt.show()
```

```python
#bar chart
plt.bar(sales,products,label="sales")
plt.title("products vs sales")
plt.xlabel("products")
plt.ylabel("amount")
plt.legend()#used to show legend
plt.show()
```

```python
plt.bar(products,sales,label="sales")
plt.title("products sales")
plt.xlabel("products")
plt.ylabel("sales")
plt.legend()#used to show legend
plt.show()
```

```python
x=np.arange(len(products))#stacked bar chart
width=0.35
plt.bar(x-width/2,sales,width,label="sales")
plt.bar(x+width/2,profit,width,label="profit")
plt.title("products vs sales")
plt.xlabel("products")
plt.ylabel("amount")
plt.legend()#used to show legend
plt.show()
```

```python
x=np.arange(len(products))
width=0.35
plt.bar(x+width/2,sales,width,label="sales")
plt.bar(x+width/2,profit,width,label="profit")
plt.title("products vs sales")
plt.xlabel("products")
plt.ylabel("amount")
plt.legend()#used to show legend
plt.show()
```

```python
import matplotlib.pyplot as plt
months=["jan","feb","mar","april","may","jun"]
monthly_sales = [100,120,150,140,180,200]
plt.plot(months,monthly_sales,marker="*")
plt.title("monthly sales")
plt.xlabel("month")
plt.ylabel("sales")
plt.show()
```

```python
import matplotlib.pyplot as plt
months=["jan","feb","mar","april","may","jun"]
monthly_sales = [100,120,150,140,180,200]
plt.plot(months,monthly_sales,marker="*")
plt.fill_between(months,monthly_sales)
plt.title("monthly sales")
plt.xlabel("month")
plt.ylabel("sales")
plt.show()
```

```python
import matplotlib.pyplot as plt
products=["laptop","mobile","tablet","watch"]
sales=[120,180,90,110]
profit=[30,40,70,20]
plt.pie(
    sales,
    labels=products,
    autopct="%.1f%%"
)
plt.title("product sales distribution")
plt.show()
```

```python
import matplotlib.pyplot as plt
products=["laptop","mobile","tablet","watch"]
sales=[120,180,90,110]
plt.pie(
    sales,
    labels=products,
    wedgeprops={"width":0.4}
)
plt.text(0,0,
         "total1500",
         ha="center",
         va="center"
)
plt.show()
```
