```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
file_path = '/content/sample_data/mnist_train_small.csv'
df = pd.read_csv(file_path)
print("Dataset loaded successfully!")
display(df.head())
display(df.tail())
print("\nDataset Shape:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("\nColumn Names:")
print(df.columns.tolist())
print(df.dtypes)
df.info()
display(df.describe())
print("\nNull values in each column:")
display(df.isnull().sum())

print("\nTotal null values:")
print(df.isnull().sum().sum())
null_summary = pd.DataFrame({
    'Null Count': df.isnull().sum(),
    'Null Percentage': (df.isnull().sum() / len(df)) * 100
})

print("\nNull Value Summary:")
display(null_summary[null_summary['Null Count'] > 0])

plt.figure(figsize=(12, 6))
sns.heatmap(df.isnull(), cbar=False)
plt.title("Missing Values Heatmap")
plt.show()
numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())

print("\nNull values after numerical imputation:")
print(df.isnull().sum().sum())

categorical_columns = df.select_dtypes(include=['object', 'category']).columns
for col in categorical_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])
print("\nNull values after categorical imputation:")
print(df.isnull().sum().sum())

duplicate_count = df.duplicated().sum()
print("\nNumber of duplicate rows:")
print(duplicate_count)

if duplicate_count > 0:
    print("\nDuplicate rows:")
    display(df[df.duplicated()])

df = df.drop_duplicates()
print("\nDuplicates after removal:")
print(df.duplicated().sum())

print("\nShape after removing duplicates:")
print(df.shape)

duplicate_columns = df.T.duplicated()
print("\nDuplicate columns:")
print(df.columns[duplicate_columns].tolist())

df = df.loc[:, ~df.T.duplicated()]
print("\nShape after removing duplicate columns:")
print(df.shape)

print("\nFinal data types:")
print(df.dtypes.value_counts())

print("\nFinal total null values:")
print(df.isnull().sum().sum())

print("\nFinal duplicate rows:")
print(df.duplicated().sum())

print("\nFinal dataset information:")
df.info()

print("\nFinal dataset shape:")
print(df.shape)

print("Total Missing Values:",df.isnull().sum().sum())
print("Total Duplicate Rows:", df.duplicated().sum())
print("Final Rows:", df.shape[0])
print("Final Columns:", df.shape[1])

cleaned_path = ('/content/sample_data/''mnist_train_small_cleaned.csv')
df.to_csv(cleaned_path,index=False)
print("\nCleaned dataset saved successfully!")
print("Path:", cleaned_path)
```

# Statistical Operations

```python
arr = np.array([10,20,30,40,50])
print(np.sum(arr)) #to print the sum of values in an array
print(np.mean(arr)) #calculates the mean of the array
print(np.max(arr)) #picks the highest value in the array
print(np.min(arr)) #picks the least value in the array
print(np.std(arr)) #calculates the Standard Deviation of the array
print(np.var(arr)) #calculates the variance of the array
print(np.median(arr)) #calculates the median of the array
print(np.percentile(arr,75)) #calculates the marks obtained based on the percentile value we passed and the array
print(np.quantile(arr,0.5)) #calculates the marks obtained based on the percentile value passed between 0-1 and on the array
print(np.argmax(arr)) #gives the position of the highest value in the array
print(np.argmin(arr)) #gives the position(0 to N-1) of the least value in the array
print(np.log10(arr)) #gives the base-10 logarithm value for every value in the array
print(np.log(arr)) #gives the base-e logarithm value for every value in the array
print(np.round(arr,2)) #prints the rounded values in the array with the int value deciding the no of decimal points it can round off to
print(np.ceil(arr)) #prints the rounded value but rounding it off to the higher value for every element in the array
print(np.floor(arr)) #prints the rounded value but rounding it off to the lower value for every element in the array
print(np.abs(arr)) #removes the negative values of in an array and prints the converted positives
print(np.sort(arr)) #prints the sorted array
print(np.sort(arr)[::-1]) #prints the sorted array but in backwards
```
